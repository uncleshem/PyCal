```python
#BMI CALCULATOR

weight = float(input("Enter your weight: "))
height = float(input("Enter your height: "))

bmi = weight/ (height ** 2)
print("BMI", bmi)

if bmi <= 18.5:
    print("Underweight")
elif bmi >= 24.9:
    print("Normal")
elif bmi >= 29.9:
    print("Overweight")
else:
    print("Obese")
```
